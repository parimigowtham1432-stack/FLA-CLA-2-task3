```python

# ================================================================
# SRM INSTITUTE OF SCIENCE AND TECHNOLOGY, TIRUCHIRAPPALLI
# FORMAL LANGUAGE AND AUTOMATA - CLA 2 | ASSIGNMENT TASK 3
# CASE STUDY-BASED ANALYSIS AND DESIGN USING PDA AND CFG
#
# CASE STUDY 1: ARITHMETIC EXPRESSION VALIDATION USING CFG
# CASE STUDY 2: BALANCED PARENTHESES VALIDATION USING PDA
# CASE STUDY 3: PALINDROME CHECKING USING NPDA
#
# STUDENT: PARIMI GOWTHAM
# REGISTER NO: RA2411026050033
# DEPARTMENT: AIML
# SEMESTER: V
# ACADEMIC YEAR: 2026-2027
# COURSE CODE: 21CSC301T
# FACULTY: Dr. C. Gunasundari
#
# RUN THIS ENTIRE CODE IN ONE GOOGLE COLAB CELL
# ================================================================

import re
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

pd.set_option("display.max_colwidth", 100)

ALL_RESULTS = []


# ================================================================
# COMMON FUNCTIONS
# ================================================================

def heading(title):
    display(Markdown("\n# " + title))


def show_text(text):
    display(Markdown(text))


def record_result(case, test_id, input_text, expected, actual, trace=None):

    status = "PASS" if expected == actual else "FAIL"

    ALL_RESULTS.append({
        "Case Study": case,
        "Test Case": test_id,
        "Input": input_text,
        "Expected": "ACCEPT" if expected else "REJECT",
        "Actual": "ACCEPT" if actual else "REJECT",
        "Status": status
    })

    print(
        f"{test_id} | Input: {input_text!r} | "
        f"Expected: {'ACCEPT' if expected else 'REJECT'} | "
        f"Actual: {'ACCEPT' if actual else 'REJECT'} | {status}"
    )

    if trace:
        for step in trace:
            print("   ", step)


def draw_diagram(title, nodes, edges):

    fig, ax = plt.subplots(figsize=(10, 4))

    ax.set_xlim(0, 10)
    ax.set_ylim(0, 4)
    ax.axis("off")

    positions = {}

    for name, x, y, accepting in nodes:

        positions[name] = (x, y)

        circle = plt.Circle(
            (x, y), 0.45,
            fill=False,
            linewidth=2
        )

        ax.add_patch(circle)

        if accepting:
            inner = plt.Circle(
                (x, y), 0.36,
                fill=False,
                linewidth=1.5
            )
            ax.add_patch(inner)

        ax.text(
            x, y, name,
            ha="center",
            va="center",
            fontsize=10
        )

    for start, end, label in edges:

        x1, y1 = positions[start]
        x2, y2 = positions[end]

        if start == end:

            ax.annotate(
                "",
                xy=(x2 - 0.2, y2 + 0.4),
                xytext=(x1 + 0.2, y1 + 0.4),
                arrowprops=dict(
                    arrowstyle="->",
                    connectionstyle="arc3,rad=-1.5",
                    linewidth=1.5
                )
            )

            ax.text(
                x1, y1 + 1,
                label,
                ha="center",
                fontsize=9
            )

        else:

            ax.annotate(
                "",
                xy=(x2 - 0.5, y2),
                xytext=(x1 + 0.5, y1),
                arrowprops=dict(
                    arrowstyle="->",
                    linewidth=1.5
                )
            )

            ax.text(
                (x1 + x2) / 2,
                (y1 + y2) / 2 + 0.25,
                label,
                ha="center",
                fontsize=9
            )

    ax.set_title(title, fontsize=13, weight="bold")

    plt.show()


# ================================================================
# CASE STUDY 1
# ARITHMETIC EXPRESSION VALIDATION USING CFG
# ================================================================

heading("CASE STUDY 1: ARITHMETIC EXPRESSION VALIDATION USING CFG")

show_text("""
## Objective

Validate arithmetic expressions using a Context-Free Grammar (CFG).

The expressions can contain:
- Identifiers
- Integer numbers
- Decimal numbers
- Addition (+)
- Subtraction (-)
- Multiplication (*)
- Division (/)
- Parentheses

## Context-Free Grammar

E  -> T E'
E' -> + T E' | - T E' | epsilon

T  -> F T'
T' -> * F T' | / F T' | epsilon

F  -> ( E ) | id | num

## Operator precedence

1. Parentheses
2. Multiplication and division
3. Addition and subtraction

The recursive-descent parser checks whether the input follows
the grammar. It does not calculate the expression's numeric value.
""")


# Tokenizer

TOKEN_RE = re.compile(
    r"\s*(?:(\d+(?:\.\d+)?)|([A-Za-z_][A-Za-z0-9_]*)|([+\-*/()]))"
)


def tokenize_arithmetic(source):

    tokens = []
    position = 0

    while position < len(source):

        if source[position:].strip() == "":
            break

        match = TOKEN_RE.match(source, position)

        if not match:
            raise ValueError(
                f"Invalid character at position {position}"
            )

        number, identifier, operator = match.groups()

        if number:
            tokens.append(("NUM", number))

        elif identifier:
            tokens.append(("ID", identifier))

        else:
            tokens.append((operator, operator))

        position = match.end()

    return tokens


class ArithmeticParser:

    def __init__(self, tokens):

        self.tokens = tokens
        self.position = 0
        self.trace = []

    def current(self):

        if self.position < len(self.tokens):
            return self.tokens[self.position]

        return None

    def consume(self, expected=None):

        token = self.current()

        if token is None:
            raise ValueError("Unexpected end of expression")

        if expected is not None and token[0] != expected:
            raise ValueError(
                f"Expected {expected}, found {token[1]}"
            )

        self.position += 1

        self.trace.append(
            f"Consumed token: {token[1]}"
        )

        return token

    # E -> T E'

    def parse_E(self):

        self.trace.append("E -> T E'")

        self.parse_T()
        self.parse_Ep()

    # E' -> + T E' | - T E' | epsilon

    def parse_Ep(self):

        token = self.current()

        if token and token[0] in ("+", "-"):

            operator = token[0]

            self.trace.append(
                f"E' -> {operator} T E'"
            )

            self.consume(operator)

            self.parse_T()
            self.parse_Ep()

        else:

            self.trace.append("E' -> epsilon")

    # T -> F T'

    def parse_T(self):

        self.trace.append("T -> F T'")

        self.parse_F()
        self.parse_Tp()

    # T' -> * F T' | / F T' | epsilon

    def parse_Tp(self):

        token = self.current()

        if token and token[0] in ("*", "/"):

            operator = token[0]

            self.trace.append(
                f"T' -> {operator} F T'"
            )

            self.consume(operator)

            self.parse_F()
            self.parse_Tp()

        else:

            self.trace.append("T' -> epsilon")

    # F -> ( E ) | id | num

    def parse_F(self):

        token = self.current()

        if token is None:
            raise ValueError("Expected operand")

        token_type, token_value = token

        if token_type == "(":

            self.trace.append("F -> ( E )")

            self.consume("(")
            self.parse_E()
            self.consume(")")

        elif token_type == "ID":

            self.trace.append("F -> id")

            self.consume("ID")

        elif token_type == "NUM":

            self.trace.append("F -> num")

            self.consume("NUM")

        else:

            raise ValueError(
                f"Invalid operand: {token_value}"
            )

    def parse(self):

        self.parse_E()

        if self.position != len(self.tokens):

            raise ValueError(
                "Unexpected token after expression"
            )

        return True


def validate_arithmetic(expression):

    try:

        tokens = tokenize_arithmetic(expression)

        if not tokens:
            return False, ["Empty expression"]

        parser = ArithmeticParser(tokens)

        parser.parse()

        return True, parser.trace

    except Exception as error:

        return False, [f"Parsing error: {error}"]


# Arithmetic test cases

arithmetic_tests = [

    ("A01", "a+b", True),
    ("A02", "a+b*c", True),
    ("A03", "(a+b)*c", True),
    ("A04", "10+20*30", True),
    ("A05", "a/b-c+d", True),
    ("A06", "((a+b)*c)", True),

    ("A07", "a+", False),
    ("A08", "*a+b", False),
    ("A09", "(a+b", False),
    ("A10", "a++b", False),
    ("A11", "a b", False),
    ("A12", "()", False)

]

print("\nARITHMETIC EXPRESSION TEST RESULTS\n")

for test_id, expression, expected in arithmetic_tests:

    actual, trace = validate_arithmetic(expression)

    record_result(
        "Arithmetic Expression CFG",
        test_id,
        expression,
        expected,
        actual,
        trace if test_id == "A03" else None
    )


show_text("""
## Leftmost Derivation

Example expression: `a+b*c`

E
=> T E'
=> F T' E'
=> id T' E'
=> id E'
=> id + T E'
=> id + F T' E'
=> id + id * F T' E'
=> id + id * id T' E'
=> id + id * id E'
=> id + id * id

The grammar separates expressions, terms and factors.
This ensures that multiplication has higher precedence than addition.

## Ambiguity

A flat grammar such as:

E -> E + E | E * E | id

can be ambiguous because it does not define operator precedence.

The revised grammar uses E, T and F to establish precedence.
""")


draw_diagram(
    "Arithmetic Expression Parser - Conceptual Stages",
    [
        ("Start", 1, 2, False),
        ("E", 3, 2, False),
        ("T", 5, 2, False),
        ("F", 7, 2, False),
        ("Accept", 9, 2, True)
    ],
    [
        ("Start", "E", "Start parsing"),
        ("E", "T", "Expression"),
        ("T", "F", "Term"),
        ("F", "Accept", "Valid factor")
    ]
)


# ================================================================
# CASE STUDY 2
# BALANCED PARENTHESES VALIDATION USING PDA
# ================================================================

heading("CASE STUDY 2: BALANCED PARENTHESES VALIDATION USING PDA")

show_text("""
## Objective

Determine whether a string contains correctly balanced parentheses.

## Target Language

L = { w | w contains balanced parentheses }

Input alphabet:

Sigma = { (, ) }

## Context-Free Grammar

S -> (S)S | epsilon

## PDA Formal Model

States:
Q = {q0, qaccept}

Input alphabet:
Sigma = {(, )}

Stack alphabet:
Gamma = {Z0, (}

Initial stack symbol:
Z0

## Stack Operations

1. When '(' is read, PUSH '(' onto the stack.
2. When ')' is read, POP '(' from the stack.
3. If ')' is read when no '(' is available, REJECT.
4. At the end, ACCEPT if only Z0 remains.

The stack remembers unmatched opening parentheses.
This allows the PDA to recognize arbitrary nesting depth.
""")


def validate_parentheses(expression):

    stack = ["Z0"]
    trace = []

    for position, symbol in enumerate(expression):

        if symbol == "(":

            stack.append("(")

            trace.append(
                f"Position {position}: Read '(' -> PUSH | Stack: {stack}"
            )

        elif symbol == ")":

            if stack[-1] == "Z0":

                trace.append(
                    f"Position {position}: Read ')' without matching '('"
                )

                return False, trace

            stack.pop()

            trace.append(
                f"Position {position}: Read ')' -> POP | Stack: {stack}"
            )

        else:

            trace.append(
                f"Invalid character: {symbol}"
            )

            return False, trace

    if stack == ["Z0"]:

        trace.append(
            "End of input: Only Z0 remains. ACCEPT."
        )

        return True, trace

    trace.append(
        f"Unmatched opening parentheses remain: {stack}"
    )

    return False, trace


parentheses_tests = [

    ("P01", "()", True),
    ("P02", "(())", True),
    ("P03", "()()", True),
    ("P04", "(()())", True),
    ("P05", "((()))", True),
    ("P06", "(", False),
    ("P07", ")", False),
    ("P08", "(()", False),
    ("P09", "())", False),
    ("P10", "(()))(", False),
    ("P11", "abc", False),
    ("P12", "", True)

]

print("\nBALANCED PARENTHESES TEST RESULTS\n")

for test_id, expression, expected in parentheses_tests:

    actual, trace = validate_parentheses(expression)

    record_result(
        "Balanced Parentheses PDA",
        test_id,
        expression,
        expected,
        actual,
        trace if test_id == "P04" else None
    )


show_text("""
## Stack Operation Example

Input: `(()())`

| Input | Operation | Stack |
|---|---|---|
| Start | Initialize | Z0 |
| ( | PUSH | Z0, ( |
| ( | PUSH | Z0, (, ( |
| ) | POP | Z0, ( |
| ( | PUSH | Z0, (, ( |
| ) | POP | Z0, ( |
| ) | POP | Z0 |
| End | ACCEPT | Z0 |

## Derivation

S
=> (S)S
=> ((S)S)S
=> (()S)S
=> (() (S)S)S
=> (()())S
=> (()())

The grammar generates balanced parentheses.
""")


draw_diagram(
    "PDA for Balanced Parentheses",
    [
        ("q0", 3, 2, False),
        ("qaccept", 8, 2, True)
    ],
    [
        ("q0", "q0", "( / PUSH"),
        ("q0", "q0", ") / POP"),
        ("q0", "qaccept", "Input finished, stack Z0")
    ]
)


# ================================================================
# CASE STUDY 3
# PALINDROME CHECKING USING NPDA
# ================================================================

heading("CASE STUDY 3: PALINDROME CHECKING USING NPDA")

show_text("""
## Objective

Check whether an input string is a palindrome over alphabet {a,b}.

A palindrome reads the same from left to right and right to left.

## Target Language

L = { w in {a,b}* | w = reverse(w) }

## Context-Free Grammar

S -> aSa
S -> bSb
S -> a
S -> b
S -> epsilon

## NPDA Working Principle

1. Read and PUSH symbols from the first half onto the stack.
2. Nondeterministically guess the midpoint.
3. For odd-length strings, consume the middle symbol.
4. Compare the remaining input with the stack.
5. POP matching symbols.
6. ACCEPT if all input is consumed and the stack is empty.

The implementation simulates possible midpoint choices.
It is an NPDA-style simulation, not a deterministic one-pass PDA.
""")


def validate_palindrome(text):

    if any(symbol not in "ab" for symbol in text):

        return False, [
            "REJECT: Input must contain only a and b."
        ]

    n = len(text)

    trace = [
        f"Input string: {text}",
        f"Input length: {n}"
    ]

    # Try every possible midpoint.
    # Also try skipping one middle symbol for odd-length strings.

    for midpoint in range(n + 1):

        for skip_middle in (False, True):

            if skip_middle and midpoint >= n:
                continue

            stack = []

            branch_trace = []

            branch_trace.append(
                f"Trying midpoint {midpoint}, "
                f"skip middle = {skip_middle}"
            )

            # Push first half.

            for symbol in text[:midpoint]:

                stack.append(symbol)

                branch_trace.append(
                    f"Read {symbol} -> PUSH | Stack: {stack}"
                )

            # Skip middle character for odd-length palindrome.

            if skip_middle:

                middle = text[midpoint]

                branch_trace.append(
                    f"Consume middle symbol: {middle}"
                )

                remaining = text[midpoint + 1:]

            else:

                remaining = text[midpoint:]

            matched = True

            # Match second half with stack.

            for symbol in remaining:

                if not stack:

                    branch_trace.append(
                        "Stack empty before input finished."
                    )

                    matched = False
                    break

                if stack[-1] == symbol:

                    stack.pop()

                    branch_trace.append(
                        f"Read {symbol} -> MATCH and POP | Stack: {stack}"
                    )

                else:

                    branch_trace.append(
                        f"Mismatch: input {symbol}, stack top {stack[-1]}"
                    )

                    matched = False
                    break

            if matched and len(stack) == 0:

                trace.extend(branch_trace)

                trace.append(
                    "A nondeterministic branch accepts the input."
                )

                return True, trace

    trace.append(
        "All possible midpoint branches rejected."
    )

    return False, trace


palindrome_tests = [

    ("N01", "a", True),
    ("N02", "b", True),
    ("N03", "aba", True),
    ("N04", "abba", True),
    ("N05", "baab", True),
    ("N06", "aab", False),
    ("N07", "abb", False),
    ("N08", "abab", False),
    ("N09", "abc", False),
    ("N10", "baa", False),
    ("N11", "", True),
    ("N12", "babb", False)

]

print("\nPALINDROME TEST RESULTS\n")

for test_id, expression, expected in palindrome_tests:

    actual, trace = validate_palindrome(expression)

    record_result(
        "Palindrome NPDA",
        test_id,
        expression,
        expected,
        actual,
        trace if test_id == "N04" else None
    )


show_text("""
## Example Derivation

Input: `abba`

S
=> aSa
=> abSba
=> abεba
=> abba

## NPDA Stack Example

Input: `abba`

1. Read a -> PUSH a
2. Read b -> PUSH b
3. Guess midpoint after ab
4. Read b -> MATCH and POP b
5. Read a -> MATCH and POP a
6. Input consumed and stack empty -> ACCEPT

For odd-length input `aba`, the NPDA pushes `a`,
consumes `b` as the middle symbol, and matches the final `a`.
""")


draw_diagram(
    "NPDA for Palindrome Recognition - Conceptual Phases",
    [
        ("qpush", 2, 2, False),
        ("qmatch", 5, 2, False),
        ("qaccept", 8.5, 2, True)
    ],
    [
        ("qpush", "qpush", "Read a/b, PUSH"),
        ("qpush", "qmatch", "Guess midpoint (epsilon)"),
        ("qmatch", "qmatch", "Matching symbol, POP"),
        ("qmatch", "qaccept", "Input finished, stack empty")
    ]
)


# ================================================================
# COMPARATIVE ANALYSIS
# ================================================================

heading("COMPARATIVE ANALYSIS")

comparison = pd.DataFrame([

    ["Target Problem",
     "Arithmetic Expression",
     "Balanced Parentheses",
     "Palindrome"],

    ["Formal Technique",
     "CFG + Recursive Descent",
     "CFG + Deterministic PDA",
     "CFG + NPDA Simulation"],

    ["Grammar",
     "E -> T E'",
     "S -> (S)S | epsilon",
     "S -> aSa | bSb | a | b | epsilon"],

    ["Stack Usage",
     "Recursive parser call stack",
     "Push and pop parentheses",
     "Push first half and pop matching second half"],

    ["Nondeterminism",
     "No",
     "No",
     "Yes, midpoint guessing"],

    ["Main Purpose",
     "Operator and operand validation",
     "Matching nested parentheses",
     "Checking reverse symmetry"]

], columns=[
    "Feature",
    "Case Study 1",
    "Case Study 2",
    "Case Study 3"
])

display(comparison)


# ================================================================
# WHY FINITE AUTOMATA ARE INSUFFICIENT
# ================================================================

heading("WHY FINITE AUTOMATA ARE INSUFFICIENT")

show_text("""
### 1. Arithmetic Expressions

Arbitrarily nested parentheses require memory of the nesting depth.
A finite automaton cannot remember unlimited nesting.

### 2. Balanced Parentheses

The machine must remember unmatched opening parentheses and match
them in the correct order. A stack is required.

### 3. Palindromes

The machine must remember the first half of an arbitrarily long
string and compare it with the second half.

### Conclusion

Finite automata have only finite state memory.
Context-Free Grammars and Pushdown Automata provide the additional
memory structure required for these context-free language problems.
""")


# ================================================================
# AMBIGUITY AND DETERMINISM ANALYSIS
# ================================================================

heading("AMBIGUITY AND DETERMINISM ANALYSIS")

show_text("""
### Case Study 1: Arithmetic Expression

The grammar separates expressions, terms and factors.

E handles addition and subtraction.
T handles multiplication and division.
F handles operands and parentheses.

This grammar establishes operator precedence and supports
deterministic recursive-descent parsing.

### Case Study 2: Balanced Parentheses

The grammar S -> (S)S | epsilon is the standard balanced-parentheses
grammar.

The PDA performs deterministic stack operations:
- PUSH when reading '('.
- POP when reading ')'.

### Case Study 3: Palindrome

The grammar S -> aSa | bSb | a | b | epsilon generates palindromes
over {a,b}.

The NPDA guesses the midpoint nondeterministically and compares
the second half with the first half using the stack.
""")


# ================================================================
# FINAL TEST REPORT
# ================================================================

heading("FINAL TEST CASE REPORT")

results_df = pd.DataFrame(ALL_RESULTS)

display(results_df)

total = len(results_df)

passed = int(
    (results_df["Status"] == "PASS").sum()
)

failed = int(
    (results_df["Status"] == "FAIL").sum()
)

success_rate = (
    passed / total * 100
    if total > 0 else 0
)

summary_df = pd.DataFrame({

    "Metric": [
        "Total Test Cases",
        "Passed",
        "Failed",
        "Success Rate"
    ],

    "Result": [
        total,
        passed,
        failed,
        f"{success_rate:.2f}%"
    ]

})

heading("OVERALL EXECUTION SUMMARY")

display(summary_df)


# ================================================================
# CASE-WISE SUMMARY
# ================================================================

heading("CASE-WISE TEST SUMMARY")

case_summary = results_df.groupby("Case Study").agg(

    Total=("Status", "count"),

    Passed=("Status", lambda x: (x == "PASS").sum()),

    Failed=("Status", lambda x: (x == "FAIL").sum())

).reset_index()

case_summary["Success Rate (%)"] = (

    case_summary["Passed"] /
    case_summary["Total"] * 100

).round(2)

display(case_summary)


# ================================================================
# TEST RESULT GRAPH
# ================================================================

heading("TEST RESULT VISUALIZATION")

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    case_summary["Case Study"],
    case_summary["Passed"],
    label="Passed"
)

ax.bar(
    case_summary["Case Study"],
    case_summary["Failed"],
    bottom=case_summary["Passed"],
    label="Failed"
)

ax.set_title(
    "Assignment 3 - Test Case Results",
    fontsize=14,
    weight="bold"
)

ax.set_xlabel("Case Studies")
ax.set_ylabel("Number of Test Cases")

ax.tick_params(
    axis="x",
    rotation=15
)

ax.legend()

plt.tight_layout()
plt.show()


# ================================================================
# EXPORT CSV REPORTS
# ================================================================

heading("GENERATING CSV REPORTS")

results_df.to_csv(
    "FLA_Task3_Test_Results.csv",
    index=False
)

comparison.to_csv(
    "FLA_Task3_Comparative_Analysis.csv",
    index=False
)

summary_df.to_csv(
    "FLA_Task3_Overall_Summary.csv",
    index=False
)

case_summary.to_csv(
    "FLA_Task3_Case_Wise_Summary.csv",
    index=False
)

print("Successfully generated:")

print("1. FLA_Task3_Test_Results.csv")
print("2. FLA_Task3_Comparative_Analysis.csv")
print("3. FLA_Task3_Overall_Summary.csv")
print("4. FLA_Task3_Case_Wise_Summary.csv")


# ================================================================
# INTERACTIVE TESTING
# ================================================================

heading("INTERACTIVE TESTING")

print("""
Select a validator:

1. Arithmetic Expression
2. Balanced Parentheses
3. Palindrome
0. Skip
""")

choice = input("Enter your choice: ").strip()

if choice == "1":

    user_input = input(
        "Enter an arithmetic expression: "
    )

    valid, trace = validate_arithmetic(user_input)

    print("\nRESULT:", "ACCEPTED" if valid else "REJECTED")

    for step in trace:
        print(step)


elif choice == "2":

    user_input = input(
        "Enter a parentheses string: "
    )

    valid, trace = validate_parentheses(user_input)

    print("\nRESULT:", "ACCEPTED" if valid else "REJECTED")

    for step in trace:
        print(step)


elif choice == "3":

    user_input = input(
        "Enter a palindrome string using a and b: "
    )

    valid, trace = validate_palindrome(user_input)

    print("\nRESULT:", "ACCEPTED" if valid else "REJECTED")

    for step in trace:
        print(step)


elif choice == "0":

    print("Interactive testing skipped.")

else:

    print("Invalid choice.")


# ================================================================
# CONCLUSION
# ================================================================

heading("CONCLUSION")

show_text(f"""
The three case studies demonstrate the use of Context-Free Grammars
and Pushdown Automata in language validation.

1. Arithmetic expressions were validated using a precedence-aware CFG.
2. Balanced parentheses were validated using deterministic stack operations.
3. Palindromes were checked using an NPDA-style midpoint simulation.

The experiments demonstrate why finite automata are insufficient
for languages requiring unbounded nesting or comparison memory.

### Final Execution Results

- Total test cases: {total}
- Passed: {passed}
- Failed: {failed}
- Success rate: {success_rate:.2f}%

Review the generated test report and any failed cases before submission.
""")

print("\n" + "=" * 70)
print("ASSIGNMENT 3 EXECUTION COMPLETED")
print("=" * 70)

print("Student: PARIMI GOWTHAM")
print("Register Number: RA2411026050033")
print("Course: 21CSC301T - Formal Language and Automata")
print("Department: AIML")
print("Semester: V")
print("Academic Year: 2026-2027")

print("\nAll code execution steps completed.")
```
